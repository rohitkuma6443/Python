## Module 2: Core Data Structures

Pandas introduces two primary data structures: `Series` and `DataFrame`. Understanding these is crucial, as they are the foundation for almost all operations in Pandas.

### Pandas Series

A `Series` is a one-dimensional labeled array capable of holding any data type (integers, strings, floats, Python objects, etc.). It's similar to a column in a spreadsheet or a SQL table, or a NumPy array with an associated index (labels).

**Creation of a Series:**

You can create a Series from a list, NumPy array, or dictionary.

```python
# Create a Series from a list
data_list = [10, 20, 30, 40, 50]

data_series = pd.Series(data_list)
print("Series from a list:")
display(data_series)

# Create a Series with a custom index
data_list_custom = ['apple', 'banana', 'cherry']
custom_index = ['a', 'b', 'c']
s2 = pd.Series(data_list_custom, index=custom_index)
print("\nSeries with custom index:")
display(s2)

# Create a Series from a dictionary
data_dict = {'A': 100, 'B': 200, 'C': 300}
s3 = pd.Series(data_dict)
print("\nSeries from a dictionary:")
display(s3)
```

### Pandas DataFrame

A `DataFrame` is a two-dimensional labeled data structure with columns of potentially different types. It is like a spreadsheet, SQL table, or a dictionary of Series objects. It is the most commonly used Pandas object.

**Creation of a DataFrame:**

You can create a DataFrame from dictionaries, lists of dictionaries, or even other DataFrames.

```python
# Create a DataFrame from a dictionary of lists
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston']
}
data_frame = pd.DataFrame(data)
print("DataFrame from a dictionary of lists:")
display(data_frame)

# Create a DataFrame from a list of dictionaries
data_list_dict = [
    {'Name': 'Eve', 'Age': 28, 'City': 'Miami'},
    {'Name': 'Frank', 'Age': 32, 'City': 'Boston'}
]
df_from_list_dict = pd.DataFrame(data_list_dict)
print("\nDataFrame from a list of dictionaries:")
display(df_from_list_dict)
```

### Data Types (dtypes)

Pandas automatically infers data types for Series and DataFrame columns. Common data types include:

You can inspect the data types using the `.dtype` (for Series) or `.dtypes` (for DataFrame) attributes.

### Reading Files

Pandas provides a suite of functions to read data from different sources. The most frequently used are `read_csv()`, `read_excel()`, and `read_json()`.

**`pd.read_csv()`: Reading from CSV files**

This is perhaps the most common function for loading tabular data. It's highly flexible and can handle various delimiters, headers, and encoding options.

```python
# Create a DataFrame for group-wise filling
df_with_na = pd.DataFrame({
    'Group': ['A', 'A', 'B', 'B', 'A'],
    'Value': [10, np.nan, 20, np.nan, 15]
})
print("Original DataFrame for group-wise filling:")
display(df_with_na)

# Fill NaN values with the mean of their respective groups
df_with_na['Value_Filled_Group_Mean'] = df_with_na.groupby('Group')['Value'].transform(lambda x: x.fillna(x.mean()))
print("\nDataFrame after filling NaN with group-wise mean:")
display(df_with_na)
```

This approach is essential for memory-efficient processing of big data.

## Module 5: Data Selection & Indexing

Data selection and indexing are fundamental operations in Pandas, allowing you to extract specific subsets of your data for analysis or manipulation. Pandas offers several powerful and flexible ways to select data, primarily through column access, row selection using labels (`.loc[]`) and integer-location (`.iloc[]`), and boolean indexing.

### 5.1 Selecting Columns

You can select one or multiple columns from a DataFrame. This returns either a Series (for a single column) or another DataFrame (for multiple columns).

**Selecting a Single Column:**

To select a single column, you can use dictionary-like notation or dot notation (if the column name is a valid Python identifier and doesn't conflict with DataFrame methods).

```python
# Select a single column (returns a Series)
names = data_frame['Name']
print("Selecting 'Name' column (Series):")
display(names)

# Alternative using dot notation (if column name allows)
ages = data_frame.Age
print("\nSelecting 'Age' column (dot notation):")
display(ages)
```

**Selecting Multiple Columns:**

To select multiple columns, pass a list of column names to the DataFrame.

```python
# Select multiple columns (returns a DataFrame)
name_age = data_frame[['Name', 'Age']]
print("\nSelecting 'Name' and 'Age' columns (DataFrame):")
display(name_age)
```

### 5.2 Selecting Rows with `.loc[]` and `.iloc[]`

Pandas provides two primary methods for selecting rows (and columns) based on their labels or integer positions:

*   **`.loc[]`**: Primarily label-based indexing. It can take a single label, a list of labels, a slice of labels, or a boolean array.
*   **`.iloc[]`**: Primarily integer-location based indexing. It can take a single integer, a list of integers, a slice of integers, or a boolean array.

**Using `.loc[]` (Label-based indexing):**

`loc` is used with row and column labels.

```python
# Select a single row by label (index)
row_0_loc = data_frame.loc[0]
print("Selecting row with label 0 using .loc:")
display(row_0_loc)

# Select multiple rows by labels
rows_0_2_loc = data_frame.loc[[0, 2]]
print("\nSelecting rows with labels 0 and 2 using .loc:")
display(rows_0_2_loc)

# Select rows and specific columns by labels
rows_cols_loc = data_frame.loc[0:2, ['Name', 'City']]
print("\nSelecting rows 0-2 and 'Name', 'City' columns using .loc:")
display(rows_cols_loc)
```

**Using `.iloc[]` (Integer-location based indexing):**

`iloc` is used with integer positions, just like standard Python indexing.

```python
# Select a single row by integer position
row_1_iloc = data_frame.iloc[1]
print("Selecting row at integer position 1 using .iloc:")
display(row_1_iloc)

# Select multiple rows by integer positions
rows_1_3_iloc = data_frame.iloc[[1, 3]]
print("\nSelecting rows at integer positions 1 and 3 using .iloc:")
display(rows_1_3_iloc)

# Select rows and specific columns by integer positions
rows_cols_iloc = data_frame.iloc[0:2, [0, 2]] # rows 0 to 1, columns 0 and 2
print("\nSelecting rows 0-1 and columns 0, 2 using .iloc:")
display(rows_cols_iloc)
```

### 5.3 Boolean Indexing (Filtering Data)

Boolean indexing, also known as boolean masking, is a powerful way to filter data based on conditions. You pass a boolean Series (or array) to the DataFrame, and only rows where the condition is `True` are returned.

**Filtering with a Single Condition:**

```python
# Filter rows where 'Age' is greater than 30
filtered_df = data_frame[data_frame['Age'] > 30]
print("DataFrame where Age > 30:")
display(filtered_df)

# Filter rows where 'City' is 'New York'
from_ny = data_frame[data_frame['City'] == 'New York']
print("\nDataFrame where City is 'New York':")
display(from_ny)
```

**Filtering with Multiple Conditions (AND / OR):**

Combine multiple conditions using `&` (AND) or `|` (OR) operators. Remember to wrap each condition in parentheses.

```python
# Multiple conditions with AND (&)
# Select people older than 30 AND living in Chicago
complex_condition = (data_frame['Age'] > 30) & (data_frame['City'] == 'Chicago')
aged_over_30 = data_frame[complex_condition]
print("DataFrame where Age > 30 AND City == 'Chicago':")
display(aged_over_30)

# Multiple conditions with OR (|)
# Select people from New York OR Chicago
complex_condition_or = (data_frame['City'] == 'New York') | (data_frame['City'] == 'Chicago')
filtered_df_or = data_frame[complex_condition_or]
print("\nDataFrame where City is 'New York' OR 'Chicago':")
display(filtered_df_or)
```

### 5.4 Using `.isin()` for Multiple Value Filtering

The `.isin()` method is very useful for filtering when you want to select rows where a column's value is one of several possible values.

```python
# Filter rows where 'City' is either 'New York' or 'Chicago'
cities_of_interest = ['New York', 'Chicago']
filtered_by_list = data_frame[data_frame['City'].isin(cities_of_interest)]
print("DataFrame where City is in ['New York', 'Chicago'] using .isin():")
display(filtered_by_list)
```

## Module 6: Data Cleaning & Preprocessing

Data cleaning and preprocessing are critical steps in any data analysis pipeline. Raw data is often messy, containing missing values, duplicates, incorrect data types, or outliers. This module will cover fundamental techniques using Pandas to address these common issues, making your data ready for analysis.

### 6.1 Handling Missing Data

Missing data is a common problem. Pandas represents missing values as `NaN` (Not a Number) for numerical columns and `None` or `NaN` for object/string columns. Effectively dealing with missing data is crucial as many analytical tools cannot handle `NaN`s.

Key methods:
*   **`isnull()` / `isna()`**: Returns a boolean DataFrame indicating where values are missing.
*   **`notnull()` / `notna()`**: Returns a boolean DataFrame indicating where values are not missing.
*   **`dropna()`**: Removes rows or columns containing missing values.
*   **`fillna()`**: Fills missing values with a specified value or using a method (e.g., forward-fill, back-fill, mean, median).

**Creating a DataFrame with Missing Values for Demonstration:**

```python
data_missing = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, 4, np.nan],
    'C': [10, 20, 30, np.nan, 50]
}
df_missing = pd.DataFrame(data_missing)
print("Original DataFrame with missing values:")
display(df_missing)
```

**Checking for Missing Values:**

```python
print("DataFrame.isnull():")
display(df_missing.isnull())

print("\nSum of missing values per column:")
display(df_missing.isnull().sum())

print("\nTotal number of missing values:")
display(df_missing.isnull().sum().sum())
```

**Dropping Missing Values:**

```python
# Drop rows with any missing values
df_dropped_rows = df_missing.dropna()
print("DataFrame after dropping rows with any NaN:")
display(df_dropped_rows)

# Drop columns with any missing values
df_dropped_cols = df_missing.dropna(axis=1)
print("\nDataFrame after dropping columns with any NaN:")
display(df_dropped_cols)

# Drop rows only if ALL values are missing (not applicable in this example)
df_dropped_all_na = df_missing.dropna(how='all')
print("\nDataFrame after dropping rows where all values are NaN (no change here):")
display(df_dropped_all_na)
```

**Filling Missing Values:**

```python
# Fill with a constant value (e.g., 0)
df_filled_zero = df_missing.fillna(0)
print("DataFrame after filling NaN with 0:")
display(df_filled_zero)

# Fill with the mean of each column
df_filled_mean = df_missing.fillna(df_missing.mean())
print("\nDataFrame after filling NaN with column means:")
display(df_filled_mean)

# Forward-fill (propagate last valid observation forward to next valid)
df_ffill = df_missing.fillna(method='ffill')
print("\nDataFrame after forward-fill:")
display(df_ffill)

# Back-fill (propagate next valid observation backward to next valid)
df_bfill = df_missing.fillna(method='bfill')
print("\nDataFrame after back-fill:")
display(df_bfill)
```

**Conditional Filling (e.g., group-wise mean):**

Sometimes, filling missing values based on groups is more appropriate.

```python
# Create a DataFrame for group-wise filling
df_with_na = pd.DataFrame({
    'Group': ['A', 'A', 'B', 'B', 'A'],
    'Value': [10, np.nan, 20, np.nan, 15]
})
print("Original DataFrame for group-wise filling:")
display(df_with_na)

# Fill NaN values with the mean of their respective groups
df_with_na['Value_Filled_Group_Mean'] = df_with_na.groupby('Group')['Value'].transform(lambda x: x.fillna(x.mean()))
print("\nDataFrame after filling NaN with group-wise mean:")
display(df_with_na)
```

### 6.2 Handling Duplicate Data

Duplicate rows can skew analysis and lead to incorrect conclusions. Pandas provides methods to identify and remove duplicate entries.

Key methods:
*   **`duplicated()`**: Returns a boolean Series indicating whether each row is a duplicate of a previous row.
*   **`drop_duplicates()`**: Removes duplicate rows.

**Creating a DataFrame with Duplicate Values for Demonstration:**

```python
data_duplicates = {
    'Name': ['Alice', 'Bob', 'Alice', 'Charlie', 'Bob'],
    'Age': [25, 30, 25, 35, 30],
    'City': ['NY', 'LA', 'NY', 'CHI', 'LA']
}
df_duplicates = pd.DataFrame(data_duplicates)
print("Original DataFrame with duplicates:")
display(df_duplicates)
```

```python
print("Boolean Series indicating duplicated rows:")
display(df_duplicates.duplicated())

print("\nDuplicated rows themselves (keeping the first occurrence):")
display(df_duplicates[df_duplicates.duplicated()])
```

**Removing Duplicates:**

```python
# Remove all duplicate rows (keeps the first occurrence by default)
df_no_duplicates = df_duplicates.drop_duplicates()
print("DataFrame after dropping all duplicate rows:")
display(df_no_duplicates)

# Remove duplicates based on a subset of columns (e.g., 'Name' and 'Age')
df_no_duplicates_subset = df_duplicates.drop_duplicates(subset=['Name', 'Age'])
print("\nDataFrame after dropping duplicates based on 'Name' and 'Age':")
display(df_no_duplicates_subset)

# Keep the last occurrence of duplicates instead of the first
df_no_duplicates_last = df_duplicates.drop_duplicates(keep='last')
print("\nDataFrame after dropping duplicates, keeping the last occurrence:")
display(df_no_duplicates_last)
```

## Module 8: Data Aggregation & Grouping

Data aggregation and grouping are fundamental operations for summarizing data and extracting meaningful insights. Pandas provides the incredibly versatile `groupby()` method, often used in conjunction with `agg()`, `pivot_table()`, and other reshaping functions.

### 8.1 Grouping with `groupby()`

The `groupby()` method is used to group data based on one or more columns, allowing you to perform aggregate functions (like sum, mean, count, min, max) on the grouped data. It follows a "split-apply-combine" strategy:
1.  **Split**: Divide the DataFrame into groups based on some criteria.
2.  **Apply**: Apply a function to each group independently.
3.  **Combine**: Combine the results into a DataFrame.

**Basic Aggregation after Grouping:**

```python
# Create a sample DataFrame for grouping
df_sales = pd.DataFrame({
    'Region': ['East', 'West', 'East', 'South', 'West', 'East', 'South'],
    'Salesperson': ['Alice', 'Bob', 'Alice', 'Charlie', 'Bob', 'David', 'Charlie'],
    'Product': ['A', 'B', 'A', 'C', 'B', 'A', 'C'],
    'Sales': [100, 150, 120, 80, 200, 90, 110],
    'Orders': [10, 15, 12, 8, 20, 9, 11]
})
print("Original Sales DataFrame:")
display(df_sales)

# Group by 'Region' and calculate the sum of 'Sales'
total_sales_region = df_sales.groupby('Region')['Sales'].sum()
print("\nTotal Sales by Region:")
display(total_sales_region)

# Group by 'Region' and calculate the count of 'Orders'
order_counts = df_sales.groupby('Region')['Orders'].count()
print("\nOrder Counts by Region:")
display(order_counts)
```

**Applying Multiple Aggregation Functions (`.agg()`):**

The `.agg()` method allows you to apply multiple aggregation functions to one or more columns simultaneously.

```python
# Group by 'Region' and get the sum, mean, and count of 'Sales'
region_sales_summary = df_sales.groupby('Region')['Sales'].agg(['sum', 'mean', 'count'])
print("Sales Summary by Region (sum, mean, count):")
display(region_sales_summary)

# Apply different aggregations to different columns
multi_col_agg = df_sales.groupby('Region').agg(
    Total_Sales=('Sales', 'sum'),
    Avg_Orders=('Orders', 'mean'),
    Num_Salespersons=('Salesperson', 'nunique') # Count unique salespersons
)
print("\nCustom Aggregations by Region:")
display(multi_col_agg)
```

**Grouping by Multiple Columns:**

You can group by more than one column to create hierarchical summaries.

```python
# Group by 'Region' and 'Salesperson' and get the max 'Sales'
max_sales_person = df_sales.groupby(['Region', 'Salesperson'])['Sales'].max()
print("Max Sales by Region and Salesperson:")
display(max_sales_person)

# To get the result as a DataFrame with grouped columns as regular columns, use .reset_index()
avg_sales_person = df_sales.groupby('Salesperson')['Sales'].mean().reset_index()
print("\nAverage Sales by Salesperson (DataFrame format):")
display(avg_sales_person)
```

### 8.2 Pivoting Data with `pivot_table()`

The `pivot_table()` function is a powerful tool to summarize and reshape data. It creates a spreadsheet-style pivot table as a DataFrame. It takes arguments for `index`, `columns`, `values`, and `aggfunc` (aggregation function).

```python
# Simple pivot table: 'Region' as index, 'Product' as columns, sum of 'Sales' as values
pivot_sum = pd.pivot_table(df_sales, values='Sales', index='Region', columns='Product', aggfunc='sum')
print("Pivot Table (Sum of Sales by Region and Product):")
display(pivot_sum)

# Pivot table with mean as aggregation function and filling NaN
pivot_basic = pd.pivot_table(df_sales, values='Sales', index='Region', columns='Product', aggfunc='mean')
print("\nPivot Table (Mean of Sales by Region and Product, NaN as default):")
display(pivot_basic)

# Fill missing values in pivot table
pivot_filled = pd.pivot_table(df_sales, values='Sales', index='Region', columns='Product', aggfunc='mean', fill_value=0)
print("\nPivot Table (Mean of Sales, NaN filled with 0):")
display(pivot_filled)

# Pivot table with multiple aggregation functions
pivot_multi_agg = pd.pivot_table(df_sales, values='Sales', index='Region', columns='Product', aggfunc=['sum', 'mean'])
print("\nPivot Table (Multiple Aggregations):")
display(pivot_multi_agg)
```

**`stack()` and `unstack()`: Pivoting with MultiIndex**

`stack()` pivots a level of the column labels to the row axis, producing a Series or DataFrame with a MultiIndex. `unstack()` does the inverse, pivoting a level of the (normally row) index to the column axis.

## Module 9: Data Merging & Joining

One of the most common tasks in data analysis is combining different datasets. Pandas provides powerful tools to merge and join DataFrames based on common columns or indices, similar to SQL join operations.

### 9.1 Concatenating DataFrames (`pd.concat()`)

`pd.concat()` is used to append DataFrames either along the row axis (`axis=0`, default) or the column axis (`axis=1`). It stacks multiple DataFrames vertically or horizontally.

```python
# Create two sample DataFrames
df_concat1 = pd.DataFrame({'A': ['A0', 'A1'], 'B': ['B0', 'B1']})
df_concat2 = pd.DataFrame({'A': ['A2', 'A3'], 'B': ['B2', 'B3']})

print("df_concat1:")
display(df_concat1)
print("\ndf_concat2:")
display(df_concat2)

# Concatenate along rows (default behavior, axis=0)
concat_rows = pd.concat([df_concat1, df_concat2])
print("\nConcatenated along rows:")
display(concat_rows)

# Concatenate along columns (axis=1)
concat_cols = pd.concat([df_concat1, df_concat2], axis=1)
print("\nConcatenated along columns:")
display(concat_cols)

# Concatenate with different columns, showing NaN for missing values
df_concat3 = pd.DataFrame({'C': ['C0', 'C1'], 'D': ['D0', 'D1']})
concat_diff_cols = pd.concat([df_concat1, df_concat3], axis=1)
print("\nConcatenated with different columns:")
display(concat_diff_cols)
```

### 9.2 Merging DataFrames (`pd.merge()`)

`pd.merge()` is Pandas' primary function for combining DataFrames based on shared columns (keys), similar to SQL's JOIN operations. You can specify the type of merge:
*   **`inner` (default):** Uses intersection of keys from both DataFrames.
*   **`left`:** Uses keys from left DataFrame, keeping all left rows.
*   **`right`:** Uses keys from right DataFrame, keeping all right rows.
*   **`outer`:** Uses union of keys from both DataFrames, keeping all rows.

```python
# Create two sample DataFrames for merging
df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40]
})
df2 = pd.DataFrame({
    'ID': [1, 3, 4, 5],
    'City': ['New York', 'Chicago', 'Houston', 'Miami'],
    'Salary': [70000, 80000, 90000, 60000]
})

print("df1:")
display(df1)
print("\ndf2:")
display(df2)

# Inner Merge (default)
inner_merge = pd.merge(df1, df2, on='ID', how='inner')
print("\nInner Merge (on ID):")
display(inner_merge)

# Left Merge
left_merge = pd.merge(df1, df2, on='ID', how='left')
print("\nLeft Merge (on ID):")
display(left_merge)

# Right Merge
right_merge = pd.merge(df1, df2, on='ID', how='right')
print("\nRight Merge (on ID):")
display(right_merge)

# Outer Merge
outer_merge = pd.merge(df1, df2, on='ID', how='outer')
print("\nOuter Merge (on ID):")
display(outer_merge)
```

### 9.3 Joining DataFrames (`.join()` method)

The `.join()` method is a convenient way to combine two DataFrames that have a common index (or one DataFrame's index with another's column). It defaults to a left join.

```python
# Create two DataFrames with common indices
df_a = pd.DataFrame({'Value_A': [10, 20]}, index=['K1', 'K2'])
df_b = pd.DataFrame({'Value_B': [30, 40]}, index=['K1', 'K3'])

print("df_a:")
display(df_a)
print("\ndf_b:")
display(df_b)

# Left Join (default for .join())
joined_df = df_a.join(df_b)
print("\nLeft Join (df_a.join(df_b)):")
display(joined_df)

# Inner Join with .join()
inner_joined_df = df_a.join(df_b, how='inner')
print("\nInner Join (df_a.join(df_b, how='inner')):")
display(inner_joined_df)

# Joining on a column rather than index
df_c = pd.DataFrame({'Name': ['Alice', 'Bob', 'Charlie'], 'Dept': ['HR', 'IT', 'Sales']}, index=[1, 2, 3])
df_d = pd.DataFrame({'Employee_ID': [1, 2, 3], 'Salary': [60000, 75000, 80000]}).set_index('Employee_ID')

print("\ndf_c:")
display(df_c)
print("\ndf_d:")
display(df_d)

# Join df_c's index with df_d's index
merged_on_index_via_join = df_c.join(df_d)
print("\nJoin df_c and df_d on their indices:")
display(merged_on_index_via_join)
```

## Module 10: Working with Time Series Data

Time series data is a crucial aspect of many data analyses, especially in finance, economics, and sensor data. Pandas provides robust tools for working with dates and times, making it easy to parse, manipulate, and analyze time-indexed data.

### 10.1 Date and Time Objects (`datetime`, `Timestamp`)

Pandas builds upon Python's built-in `datetime` module and extends it with its own `Timestamp` object, which is the type used for entries in a `DatetimeIndex`.

**Converting to Datetime:**

As seen in Module 6, `pd.to_datetime()` is the primary function for converting various date-like objects (strings, integers) into `datetime` objects.

```python
# Using df_datetime_sample from Module 6 (ensure 'OrderDate' is datetime dtype)
print("DataFrame with 'OrderDate' as datetime dtype:")
display(df_datetime_sample.info())
display(df_datetime_sample.head())
```

### 10.2 DatetimeIndex and Resampling

Setting a `DatetimeIndex` is essential for leveraging Pandas' time series functionalities, such as easy slicing, resampling, and time-based calculations.

**Setting DatetimeIndex:**

```python
# Set 'OrderDate' as the index
df_time_indexed = df_datetime_sample.set_index('OrderDate')
print("DataFrame with DatetimeIndex:")
display(df_time_indexed.info())
display(df_time_indexed.head())
```

**Accessing Time-based Components:**

Once you have `datetime` objects or a `DatetimeIndex`, you can easily access various components like year, month, day, day of week, etc.

```python
# Add new columns for time-based components
df_time_indexed['OrderYear'] = df_time_indexed.index.year
df_time_indexed['OrderMonth'] = df_time_indexed.index.month
df_time_indexed['DayOfWeek'] = df_time_indexed.index.day_name()

print("DataFrame with new time components:")
display(df_time_indexed.head())
```

**Time-based Slicing:**

With a `DatetimeIndex`, you can easily slice data using dates or date ranges.

```python
# Select data for a specific year
print("Orders in 2023:")
display(df_time_indexed.loc['2023'])

# Select data for a specific month
print("\nOrders in January 2023:")
display(df_time_indexed.loc['2023-01'])

# Select data for a date range
print("\nOrders between Jan 16 and Feb 02:")
display(df_time_indexed.loc['2023-01-16':'2023-02-02'])
```

**Resampling Time Series Data:**

Resampling is the process of aggregating time series data into a lower or higher frequency. This is incredibly useful for analyzing trends over different time periods (e.g., daily to monthly, or hourly to daily).

Common resampling frequencies:
*   `'D'`: Day
*   `'W'`: Week
*   `'M'`: Month end
*   `'Q'`: Quarter end
*   `'A'`: Year end

```python
# Resample to monthly frequency and calculate the sum of 'Amount'
monthly_sales = df_time_indexed['Amount'].resample('M').sum()
print("Monthly Sales (resampled and summed):")
display(monthly_sales)
```

### 10.3 Time Deltas (`Timedelta`)

A `Timedelta` represents a duration or a difference between two `datetime` objects. They are useful for calculating durations, adding/subtracting time to/from dates.

```python
from datetime import timedelta

# Create two Timestamp objects
date1 = pd.Timestamp('2023-01-15')
date2 = pd.Timestamp('2023-01-20')

# Calculate the difference (Timedelta)
time_diff = date2 - date1
print("Time difference (Timedelta):")
display(time_diff)

# Add a Timedelta to a date
future_date = date1 + pd.Timedelta(days=7)
print("\nDate plus 7 days:")
display(future_date)

# You can also create Timedeltas directly
days_ago = pd.Timestamp.today() - pd.Timedelta(weeks=2)
print("\nTwo weeks ago from today:")
display(days_ago)
```
